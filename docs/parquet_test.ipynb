```python
from datetime import datetime

from rust_ephem import ParquetEphemeris
```

```python
begin = datetime(2029, 1, 1)
end = datetime(2029, 1, 15)
stepsize = 60
```

```python
%%time

eph = ParquetEphemeris(
    source="../../nyx_propagator/LazuliSat_Nyx_120sec_*.parquet",
    time_col="utc_iso",
    pos_cols=("x_m", "y_m", "z_m"),
    vel_cols=("vx_m_s", "vy_m_s", "vz_m_s"),
    begin=begin,
    end=end,
    step_size=stepsize,
    position_unit="m",
    velocity_unit="m/s",
)
```

```python
import matplotlib.pyplot as plt

plt.plot(eph.gcrs.cartesian.x, eph.gcrs.cartesian.y)
plt.show()
```
